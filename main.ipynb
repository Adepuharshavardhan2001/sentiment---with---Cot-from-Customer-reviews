```python
import pandas as pd

reviews = data[0]["data"]["productReviewDetails"]["reviews"]["details"]

rows = []

for r in reviews:
    rows.append({
        "customer_name": r["summary"]["secondary"],
        "rating": r["summary"]["primary"],
        "review": r["review"]["text"],
        "stay_info": r["tripSummary"]["text"]
    })

df = pd.DataFrame(rows)

print(df.head())
```

```python
results = []

for i, review in enumerate(df["review"], start=1):

    prompt = cot_prompt.replace("{review}", str(review))

    response = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[
            {"role": "user", "content": prompt}
        ]
    )

    output = response.choices[0].message.content

    print(f"\n{'='*50}")
    print(f"Review {i}")
    print(f"{'='*50}")
    print(output)

    results.append(output)

df["CoT_Output"] = results
```

# choosing right paramters using Tree-of-thought

```python

import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV
```

```python
data = load_breast_cancer()

X = data.data
y = data.target
```

```python
param_grid = {
    "n_estimators":[50,100,150],
    "max_depth":[3,5,10],
    "min_samples_split":[2,5]
}
```

```python
rf = RandomForestClassifier(random_state=42)

grid = GridSearchCV(
    rf,
    param_grid,
    cv=5,
    scoring="accuracy",
    return_train_score=True
)

grid.fit(X,y)
```

```python
type(results)
```

```python
import pandas as pd

tuning_results = pd.DataFrame(grid.cv_results_)

tuning_results = tuning_results[
[
    "param_n_estimators",
    "param_max_depth",
    "param_min_samples_split",
    "mean_test_score",
    "mean_train_score",
    "mean_fit_time"
]
]

tuning_results["overfit_gap"] = (
    tuning_results["mean_train_score"]
    -
    tuning_results["mean_test_score"]
)

tuning_results.head()
```

```python
tuning_results = tuning_results.sort_values(
    by="mean_test_score",
    ascending=False
)

tuning_results.head(10)
```

```python
results_text = tuning_results.to_string(index=False)
```

```python
tot_prompt = f"""
You are a Machine Learning expert.

Analyze the Random Forest hyperparameter tuning results using Tree-of-Thought reasoning.

Structure your answer exactly as follows:

## Thought Branch A (max_depth = 3)

For each configuration:
- Validation Accuracy
- Training Time
- Overfitting Gap
- Bias-Variance Analysis

Then summarize:
- Strengths
- Weaknesses
- Best candidate in Branch A

## Thought Branch B (max_depth = 5)

For each configuration:
- Validation Accuracy
- Training Time
- Overfitting Gap
- Bias-Variance Analysis

Then summarize:
- Strengths
- Weaknesses
- Best candidate in Branch B

## Thought Branch C (max_depth = 10)

For each configuration:
- Validation Accuracy
- Training Time
- Overfitting Gap
- Bias-Variance Analysis

Then summarize:
- Strengths
- Weaknesses
- Best candidate in Branch C

## Final Comparison

Create a comparison table containing:
- Configuration
- Validation Accuracy
- Overfitting Gap
- Training Time

Compare the best candidate from each branch.

## Final Decision

Clearly state:

Best Hyperparameter Configuration:
- n_estimators = ?
- max_depth = ?
- min_samples_split = ?

Justification:
- Why this configuration was selected
- Why other candidates were rejected
- Discussion of accuracy, overfitting risk, and computational cost

Hyperparameter Results:

{results_text}
"""
```

```python
response = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {"role":"user","content":tot_prompt}
    ]
)

print(response.choices[0].message.content)
```
